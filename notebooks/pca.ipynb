## PCA for GWAS

```python
import pandas as pd
import plotly.express as px
```

```python
def plot_pcs(df, color="Group"):
    fig = px.scatter(
        df,
        x="PC1",
        y="PC2",
        color=color,
        hover_name="IID",
        title="PCA of samples in GWAS analysis",
    )
    fig.update_layout(
        plot_bgcolor="rgba(0, 0, 0, 0)",
        paper_bgcolor="rgba(0, 0, 0, 0)",
        font=dict(size=24),
        height=800,
        width=800,
    )
    fig.update_traces(marker=dict(size=5))

    fig.update_xaxes(showline=True, linewidth=2, linecolor="black", mirror=True)
    fig.update_yaxes(showline=True, linewidth=2, linecolor="black", mirror=True)
    fig.update_layout(legend=dict(x=1, y=1, xanchor="right", yanchor="top"))
    fig.write_html(
        f"C:\\Users\\wdecoster\\OneDrive - Universiteit Antwerpen\\Work\\Writing\\2025_golga8a\\PCA_FTLD-FUS_allCTRL_{color}.html"
    )

def determine_haplotype(row):
    if row["delT"] > 0:
        return "HaplotypeA"
    elif row["625"] > 0:
        return "HaplotypeB"
    else:
        return "Other"

df = pd.read_csv("C:\\Users\\wdecoster\\OneDrive - Universiteit Antwerpen\\Work\\Writing\\2025_golga8a\\PCs_FTLD-FUS_allCTRL.txt", sep=" ")
variants = pd.read_excel(
    "C:\\Users\\wdecoster\\OneDrive - Universiteit Antwerpen\\Work\\Writing\\2025_golga8a\\tagging_snv_genotypes.xlsx",
    usecols=["sample", "delT", "625"],
).rename(columns={"sample": "IID"})
df = df.merge(variants, on="IID", how="left")
df["delT"] = df["delT"].apply(lambda x: -1 if x == "missing" else x)
df["625"] = df["625"].apply(lambda x: -1 if x == "missing" else x)
df["Haplotype"] = df.apply(lambda row: determine_haplotype(row), axis=1)

df["Group"] = (
    df["CC"].apply(lambda x: "aFTLD-U" if x == 2 else "CTRL").astype("category")
    )
```

```python
plot_pcs(df, color="Group")
```

```python
plot_pcs(df, color="Haplotype")
```
